**pie chart**

- autopct= %a.bf%%
- first % is the formatting
- a: how much width
- b: after point how many decimals
- %%: % symbol
- %1.1f%%: 40.2%
- %1.2f%%: 40.23%

```python
keys=visa_df['continent'].value_counts().keys()
values=visa_df['continent'].value_counts().values
plt.pie(values)
```

```python
keys=visa_df['continent'].value_counts().keys()
values=visa_df['continent'].value_counts().values
plt.pie(values)
plt.show()
```

```python
keys=visa_df['continent'].value_counts().keys()
values=visa_df['continent'].value_counts().values
plt.pie(values,labels=keys)
plt.show()
```

```python
keys=visa_df['continent'].value_counts().keys()
values=visa_df['continent'].value_counts().values
plt.pie(values,
        labels=keys,
       autopct='%1.2f%%',
       startangle=40)
plt.show()
```

```python
keys=visa_df['continent'].value_counts().keys()
values=visa_df['continent'].value_counts().values
plt.pie(values,
        labels=keys,
       autopct='%1.2f%%',
       startangle=180)
plt.show()
```

```python
keys=visa_df['continent'].value_counts().keys()
values=visa_df['continent'].value_counts().values
plt.pie(values,
        explode=[0.1 for i in range(6)],
       labels=keys,
       autopct='%1.2f%%',
       startangle=40)
plt.show()
```

```python
keys=visa_df['continent'].value_counts().keys()
values=visa_df['continent'].value_counts().values
plt.pie(values,
        explode=[0.1 for i in range(6)],
       labels=keys,
       autopct='%1.2f%%',
        radius=0.5,
       startangle=40)
plt.savefig('Continent_pie.jpg')
plt.show()
```

```python
colors = ['red', 'blue', 'green', 'yellow', 'orange', 'purple']
keys=visa_df['continent'].value_counts().keys()
values=visa_df['continent'].value_counts().values
plt.pie(values,
        explode=[0.1 for i in range(6)],
       labels=keys,
        colors=colors,
       autopct='%1.2f%%',
        radius=0.5,
       startangle=40)
plt.show()
```
