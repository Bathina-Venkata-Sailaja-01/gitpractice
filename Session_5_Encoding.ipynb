**pd.get_dummies**

- get dummies id a way to represent the categorical columns to numerical only
- first it wil check what are the labels in a column
- for example case status has two labels
     - Denied and Certified
- so it will create two new columns based on albels
     - col1: case_status_Denied
     - col2: case_status_certified
- both columns are indepent to each other that means 90 degrees.

case_status   |  case_status_Denied  | case_status_Certified 
 Denied       |         1            |           0           
 Certified    |         0            |           1           
 Certified    |         0            |           1           

- **It is also called as one hot encoder**
- it makes new columns are independent to each othe
- some of ML algorithm expect that columns should be **independent** each other and those 2 columns don't have relation.
