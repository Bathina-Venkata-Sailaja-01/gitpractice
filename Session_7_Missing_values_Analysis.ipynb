**KNNImputer**

- KNN: K-Nearest neighbours
- where K is Number
- instead of taking avg of all the values data
     - first we will find the neighbours who are close with missing data
     - by using Euclidian distance calculation will identify the neighbour rows first
     - suppose k=2 will identify the 2 closest point first
     - those closest rows values will take it, will calculate the mean
     - Finally will fill with the mean

```python
salary=[50000,np.nan,70000,90000]
age=[25,28,30,40]
df=pd.DataFrame(zip(salary,age),
                columns=['Salary','Age'])
df
```

- Sklearn
   - impute
     - KNNImpute

```python
from sklearn.impute import KNNImputer
imputer=KNNImputer(n_neighbors=2)
pd.DataFrame(imputer.fit_transform(df))
```
